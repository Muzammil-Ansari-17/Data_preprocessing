# Simple Imputer

It replaces missing values (like NaNs) with a chosen statistical measure,
such as the column's mean, median, most frequent value, or a constant.

For Numerical values


```python
import pandas as pd
import numpy as np
from pandas.conftest import axis

data = {
    "Age": [22, 25, np.nan, 30, 28, np.nan, 35, 24],
    "Salary": [40000, 50000, 45000, np.nan, 60000, 52000, np.nan, 48000],
    "Experience": [1, 3, 2, 5, np.nan, 4, 7, np.nan],

}

df = pd.DataFrame(data)

print(df)
```

```python
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy='median')
df  = pd.DataFrame(
    imputer.fit_transform(df),
    columns=df.columns
)

print(df)
```

For Categorical Values

```python
import pandas as pd
import numpy as np

data = {
    "City": ["Karachi", "Lahore", np.nan, "Karachi", "Islamabad",
             "Lahore", np.nan, "Karachi"],
    "Purchased": ["Yes", "No", "Yes", "No", np.nan, "Yes", "No", "Yes"]
}

print(df.isnull().sum())

df = pd.DataFrame(data)

print(df)
```

```python
from sklearn.impute import SimpleImputer

C_imputer = SimpleImputer(strategy='most_frequent')

df = pd.DataFrame(
    C_imputer.fit_transform(df),
    columns=df.columns
)

print(df)
```

When both numerical and categorical comes in a single dataset

```python
data = {
    "Age": [22, 25, np.nan, 30],
    "Salary": [40000, np.nan, 50000, 60000],
    "City": ["Karachi", "Lahore", np.nan, "Karachi"]
}

for key,value  in data.items():
    print(key,value)

df = pd.DataFrame(data)
```

```python
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer

num_cols = ['Age','Salary']
cat_cols = ['City']

preprocessor = ColumnTransformer(
    [
        ('num_imputer',SimpleImputer(strategy='median'),num_cols),
        ('cat_imputer',SimpleImputer(strategy='most_frequent'),cat_cols)
    ]
)

result = preprocessor.fit_transform(df)

print(result)
```

# One Label-Encoder

LabelEncoder is a scikit-learn utility that converts categorical text labels into numerical integer values (e.g., 0, 1, 2) within a single column.

```python
import pandas as pd

data = {
    "Study_Hours": [2, 5, 3, 7, 4, 6],
    "Result": ["Fail", "Pass", "Fail", "Pass", "Fail", "Pass"]
}

df = pd.DataFrame(data)

```

```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()
df['Result'] = label_encoder.fit_transform(df['Result'])

print(df)
```

# One Hot-Encoder

OneHotEncoder is a scikit-learn preprocessing tool that converts categorical data (like text labels) into a binary matrix format.

It creates a new binary column (containing 0s or 1s) for each unique category, preventing machine learning models from incorrectly assuming a natural ranking or mathematical order among the categories.

```python
import pandas as pd

data = {
    "Age": [22, 25, 30, 28, 11, 80],
    "City": ["Karachi", "Lahore", "Karachi", "Islamabad","Peshawar","Quetta"]
}

for keys,name in data.items():
    print(keys,name)

df = pd.DataFrame(data)
```

```python
from sklearn.preprocessing import OneHotEncoder

hot_encoder = OneHotEncoder(sparse_output=False)

encoded = hot_encoder.fit_transform(df[['City']])

df[hot_encoder.get_feature_names_out(['City'])]  = encoded

df.drop(columns=['City'],inplace=True)

print(df)
```

# Standard Scalar(Standardization)

Centers the average at zero and evens the spread.

Best Used When...
Data has no strict limits, and no massive outliers.


```python
import numpy as np
from sklearn.preprocessing import StandardScaler

data = np.array([
              [25,50000],
              [35,75000],
              [45,1000000]])

scalar = StandardScaler()

X_scaled = scalar.fit_transform(data)

print("Original data:\n",data)
print("Scaled data:\n",X_scaled)
```

# Min-Max Scaler (Normalization)

Squishes data into a strict, rigid box.

Best Used When...
Data has absolute natural limits (like 0-100 test marks).

```python
import  numpy as np
from sklearn.preprocessing import MinMaxScaler

data = np.array([[2,45],
                 [6,78],
                 [12,92]
                 ])

scalar = MinMaxScaler()

X_scaled = scalar.fit_transform(data)

print("orignal data : \n",data)
print("Scaled data: \n",X_scaled)
```

# Robust Scaler(Robust Scaling)

Balances data using the middle number, ignoring extremes.

Best Used When...Data contains massive, extreme outliers (like a billionaire).


```python
import numpy as np
from sklearn.preprocessing import RobustScaler

data = np.array([[50000],
                 [60000],
                 [70000],
                 [800000000]])

scalar = RobustScaler()
X_scaled = scalar.fit_transform(data)

print("Original Data: \n",data)
print("Scaled Data: \n",X_scaled)
```
